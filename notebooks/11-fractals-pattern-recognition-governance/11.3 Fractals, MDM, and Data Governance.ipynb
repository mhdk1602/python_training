## A scale-sensitive governance framework

### Candidate enterprise objects

- entity-resolution graphs
- lineage graphs
- product and customer hierarchies
- issue-propagation trees
- missingness or exception heatmaps across fields, records, and domains

### Working propositions

1. defect patterns can recur across field, record, entity, domain, and system scales
2. governance risk is often concentrated at boundaries: merges, mappings, schema transitions, and hierarchy joins
3. MDM provides the stabilizing identity layer for those boundaries
4. fractal descriptors may help rank instability, but only within bounded and testable object classes

### Measurement ideas

- box-counting dimension for binary issue maps or graph coverings
- lacunarity for uneven missingness or branch sparsity
- Hurst-style features for issue intensity over time
- local graph dimension for duplicate-cluster topology


```python
import pandas as pd

issues = pd.DataFrame(
    [
        {"scale": "field", "object": "customer.email", "issue": "missing", "count": 84},
        {"scale": "field", "object": "customer.country", "issue": "code drift", "count": 28},
        {"scale": "record", "object": "customer record", "issue": "duplicate", "count": 41},
        {"scale": "entity", "object": "customer cluster", "issue": "survivorship conflict", "count": 12},
        {"scale": "domain", "object": "customer hierarchy", "issue": "orphan node", "count": 7},
        {"scale": "system", "object": "crm to erp mapping", "issue": "reference mismatch", "count": 9},
    ]
)

issues.groupby(["scale", "issue"], as_index=False)["count"].sum()
```

The toy table above is not a fractal analysis. It is a staging move. It shows how governance issues can be organized across scale before any scale-sensitive estimator is applied.

## Where this can help in practice

- **MDM triage**: focus stewardship on clusters whose ambiguity persists across merge thresholds or hierarchy levels
- **hierarchy governance**: measure irregular branching and sparse subtrees in product or customer structures
- **lineage risk**: identify mappings or transformations where local defects propagate far downstream
- **issue management**: distinguish isolated defects from repeating cross-scale patterns

## Where this can go wrong

- treating metaphor as proof
- estimating a dimension without testing scale stability
- using one descriptor as if it settles governance decisions
- ignoring organizational ownership and escalation because the math looks elegant
